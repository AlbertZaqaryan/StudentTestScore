```python
X = df_train.drop('exam_score', axis=1)
y = df_train['exam_score']
X_test = df_test.copy()
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
model = SGDRegressor()
pipe = Pipeline([('scaler', scaler), ('model', model)])
```

```python
params = {
    # 'model__loss':['squared_error', 'huber', 'epsilon_insensitive'],
    'model__penalty' : ['l2', 'l1', 'elasticnet', None],
    'model__alpha':[0.001, 0.01, 0.1, 1],
    'model__eta0':[0.001, 0.01, 0.1, 1],
    
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_model = GridSearchCV(pipe, params, cv=3, verbose=2)
```

```python
grid_model.fit(X, y)
```

```python
y_pred = grid_model.predict(X_test)
```

```python
df_sub = pd.read_csv('sample_submission.csv')
df_sub['exam_score'] = y_pred
df_sub.to_csv('sub.csv', index=False, index_label=False)
```
